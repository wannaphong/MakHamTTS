```python
from datasets import config

# Force the datasets library to ignore torchcodec and fallback to soundfile
config.TORCHCODEC_AVAILABLE = False
from datasets import load_dataset, Audio

# 1. Load the dataset from your CSV file
# Change 'train' to whatever split makes sense, or omit it to load as a DatasetDict
dataset = load_dataset("csv", data_files="data_use.csv", split="train")

# 2. (Optional but Recommended) Rename the column
# Most HF models expect the audio column to be named simply 'audio'
dataset = dataset.rename_column("audio_filepath", "audio")
```

```python
dataset = dataset.cast_column("audio", Audio(decode=True))
```

```python
import os
import pickle
```

```python
sound_dir="filter2_dir"
```

```python
with open("cc.pkl", "rb") as f:
    data=pickle.load(f)
```

```python
len(data)
```

```python
data[-1]
```

```python
list_txt=[]
list_index=[]
for fname in os.listdir(sound_dir):
    if not fname.endswith(".wav"):
        continue
    idx = int(os.path.splitext(fname)[0])
    list_txt.append(str(data[idx]))
    list_index.append(str(idx)+".wav")
```

```python
import pandas as pd
df = pd.DataFrame({
    "index":list_index,
    "txt":list_txt
})
```

```python
df
```

```python
df.to_csv("data_ok.csv",index=False)
```

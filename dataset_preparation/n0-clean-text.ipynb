```python
import re

def is_thai_only(text):
    # Ensure the text is not empty or just whitespace
    if not text or text.isspace():
        return False
        
    # Regex pattern: 
    # \u0e00-\u0e7f covers the entire Thai Unicode block
    # \s allows for spaces between words
    thai_pattern = r'^[\u0e00-\u0e7f\s]+$'
    
    return bool(re.match(thai_pattern, text))

def check_thai(t):
    return is_thai_only(t) and "ๆ" not in t
```

```python
def read_file(p):
    with open(p,"r") as f:
        l=[i.strip() for i in f.readlines() if check_thai(i) and len(i.strip())>1]
    return l
```

```python
import glob
```

```python
list_file=list(glob.glob("text/*.txt"))
list_sent=[]
for i in list_file:
    list_sent+=read_file(i)
```
